```python
sinllama_drift_report = pd.read_csv("../Data/SinLlama_lemma_drift_analysis.csv")
xlmr_drift_report = pd.read_csv("../Data/XLM-R_lemma_drift_analysis.csv")
llama_ft_drift_report = pd.read_csv("../Data/LLaMA_FT_lemma_drift_analysis.csv")
```

```python
sin_keys = sinllama_drift_report['lemma'].unique().tolist()
xlm_keys = xlmr_drift_report['lemma'].unique().tolist()
llama_ft_keys = llama_ft_drift_report['lemma'].unique().tolist()
```

```python
xlmr_info = {'Avg Global Distance': xlmr_drift_report['global_distance'].mean(),
             'Avg Persistent Shift': xlmr_drift_report['persistent_shift'].mean(),
             'Avg Drift Reduction %': xlmr_drift_report['drift_reduction_pct'].mean(),
             'Drifted Instances': xlmr_drift_report['innovation_count'].sum(),
             'Total Instances': xlmr_drift_report['total_instance_count'].sum()}
```

```python
sinllama_info = {'Avg Global Distance': sinllama_drift_report['global_distance'].mean(),
             'Avg Persistent Shift': sinllama_drift_report['persistent_shift'].mean(),
             'Avg Drift Reduction %': sinllama_drift_report['drift_reduction_pct'].mean(),
             'Drifted Instances': sinllama_drift_report['innovation_count'].sum(),
             'Total Instances': sinllama_drift_report['total_instance_count'].sum()}
```

```python
llama_ft_info = {'Avg Global Distance': llama_ft_drift_report['global_distance'].mean(),
             'Avg Persistent Shift': llama_ft_drift_report['persistent_shift'].mean(),
             'Avg Drift Reduction %': llama_ft_drift_report['drift_reduction_pct'].mean(),
             'Drifted Instances': llama_ft_drift_report['innovation_count'].sum(),
             'Total Instances': llama_ft_drift_report['total_instance_count'].sum()}
```

```python
sinllama_drift_report['global_distance'].mean()/xlmr_drift_report['global_distance'].mean()
```

```python
df_drift = pd.DataFrame([xlmr_info, sinllama_info, llama_ft_info], index=['XLM-R', 'SinLlama', 'LLaMA-FT'])

# 4. Clean up formatting (optional but recommended for mixed metrics)
# This keeps 4 decimal places for the averages but integers for counts
format_mapping = {
    'Avg Global Distance': '{:.2f}',
    'Avg Persistent Shift': '{:.2f}',
    'Avg Drift Reduction %': '{:.2f}%',
    'Drifted Instances': '{:,.0f}',
    'Total Instances': '{:,.0f}',
    'Total': '{:.2f}'
}

# Apply formatting for display
print(df_drift.style.format(format_mapping))
```
