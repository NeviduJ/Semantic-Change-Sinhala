```python
import pandas as pd
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
sim_mat_report_wv = pd.read_csv("../Data/Word2Vec_dip_neighbors_similarity_matrix.csv")
analytical_op_report_wv = pd.read_csv("../Data/word2vec_dip_neighbors_orthogonal_procrustes.csv")
neural_op_report_wv = pd.read_csv("../Data/word2vec_dip_neighbors_orthogonal_procrustes_NN.csv")

sim_mat_report_ft = pd.read_csv("../Data/fasttext_dip_neighbors_similarity_matrix.csv")
analytical_op_report_ft = pd.read_csv("../Data/fasttext_dip_neighbors_orthogonal_procrustes.csv")
neural_op_report_ft = pd.read_csv("../Data/fasttext_neighbors_orthogonal_procrustes_NN.csv")
```

```python
len(sim_mat_report_wv), len(analytical_op_report_wv), len(neural_op_report_wv), len(sim_mat_report_ft), len(analytical_op_report_ft), len(neural_op_report_ft)
```

```python
counts_data_wv = {
    'SMA': sim_mat_report_wv['Period'].value_counts(),
    'Analytical OP': analytical_op_report_wv['Period'].value_counts(),
    'Neural OP': neural_op_report_wv['Period'].value_counts()
}
```

```python
counts_data_ft = {
    'SMA': sim_mat_report_ft['Period'].value_counts(),
    'Analytical OP': analytical_op_report_ft['Period'].value_counts(),
    'Neural OP': neural_op_report_ft['Period'].value_counts()
}
```

```python
df_wv = pd.DataFrame(counts_data_wv).T
df_ft = pd.DataFrame(counts_data_ft).T

df_wv.index = pd.MultiIndex.from_product([['Word2Vec'], df_wv.index], names=['Model', 'Approach'])
df_ft.index = pd.MultiIndex.from_product([['FastText'], df_ft.index], names=['Model', 'Approach'])

df_merged = pd.concat([df_wv, df_ft]).fillna(0).astype(int)

df_merged['Total'] = df_merged.sum(axis=1)

df_merged.loc[('Total', ''), :] = df_merged.sum(axis=0)
```
